```python
# ==========================================
# NEUROTRADE MASTER LAUNCHER
# ==========================================
import os
import subprocess
import time
import glob

# 1. Setup Environment
print("🛠️ Setting up environment...")
REPO_URL = "https://github.com/aryan12375/neurotrade-lstm"

if not os.path.exists('app'):
    !git clone {REPO_URL} temp_repo
    !cp -r temp_repo/* .
    !cp -r temp_repo/.* .
    !rm -rf temp_repo

# Fix Python Path
os.environ['PYTHONPATH'] += f":{os.getcwd()}"

if os.path.exists('requirements.txt'):
    !pip install -q -r requirements.txt
    !npm install -g localtunnel
    print("✅ Dependencies installed.")

# 2. Train Models
print("\n📈 Training LSTM Models (This might take a few minutes)...")
!python scripts/train.py --all

# 3. Setup Frontend for Cloud Deployment
print("\n🖥️ Setting up Dashboard...")
# Find the HTML file (handles both fixed and standard names)
html_files = glob.glob('lstm*.html')
if html_files:
    html_file = html_files[0]
    with open(html_file, 'r', encoding='utf-8') as f:
        html = f.read()

    # IMPORTANT: Change 'localhost' API calls to relative paths so it works over the tunnel
    html = html.replace('http://localhost:8000', '')

    # Move to 'frontend/' so FastAPI automatically serves it!
    os.makedirs('frontend', exist_ok=True)
    with open('frontend/index.html', 'w', encoding='utf-8') as f:
        f.write(html)
    print(f"✅ Dashboard linked.")

# 4. Start Server & Tunnel
print("\n🚀 Starting NeuroTrade API & Dashboard...")
subprocess.Popen(["python", "run.py", "--port", "8000"])
time.sleep(5)

print("\n✨ NEUROTRADE LIVE AT THE URL BELOW:")
!lt --port 8000
```
